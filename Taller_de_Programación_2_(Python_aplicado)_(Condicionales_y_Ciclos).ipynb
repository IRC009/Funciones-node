**1. Sumar ventas diarias**

```python
#Pida al usuario que ingrese las ventas de cada dia de la semana y muestre el total vendido al final
dias = ["Lunes", "Martes", "Miercoles", "Jueves", "Viernes", "Sabado", "Domingo"]
Ventas = [20]
for d in dias:
  valor = float(input("Venta del " + d + ": "))
  Ventas.append(valor)
  print("Ventas totales: ", sum(Ventas))
```

**2. Contar productos por categoria**

```python
#Solicitud ingresar 10 producto y sus categorias, al final mostrar cuantos productos hay por categoria
productos = []

for i in range(10):
    nombre = input("Producto #" + str(i+1) + ": ")
    categoria = input("Categoría: ")
    productos.append([nombre, categoria])  # <-- aquí está la corrección

conteo = {}

for prod in productos:
    cat = prod[1]  # la categoría está en la posición 1
    if cat in conteo:
        conteo[cat] += 1
    else:
        conteo[cat] = 1

print("Conteo de productos por categoría:", conteo)
```

**3. Cálculo de sueldos de empleados**

**4. Promedio de satisfacción del cliente**

**5. Inventario repetitivo**

```python
#Permite ingresar productos hasta que el usuario diga "fin", muestras cuanto productos se registraron
productos = []
while True:
  producto = input("Producto #" + str(len(productos)+1) + ": ")
  if producto == "fin":
    break
  productos.append(producto)
print("Productos registrados: ", len(productos))
```

**6. Pagos mensuales de préstamos**

```python
 #Simula el pago mensual de un prestamo, restando la cuota fija de cada mes hasta que pague completamente
monto = float(input("Monto del préstamo: "))
cuota = float(input("Cuota fija mensual: "))
saldo = monto

while saldo > 0:
    if saldo >= cuota:
        saldo = saldo - cuota
    else:
        saldo = 0
    print("Saldo restante:", saldo)
```

**9. Registrar asistencia diaria**

**10. Registro de facturas**

**11. Simulación de crecimiento de inversión**

```python
#Simula una inversion con interes compuesto por 12 meses
capital = float(input("Capital inicial: "))
interes = float(input("Interes mensual (ejemplo 0.02 para %): "))
valores = [capital]
for i in range(12):
  capital = capital * (1 + interes)
  valores.append(capital)
print(valores)
```

**15. Control de horas trabajadas**

```python
#Permite ingresar horas trabajadas por cada empleado y calcula cuantos trabajaron horas extras (>40)
n = int(input("Numero de empleados: "))
horas = []
for i in range(n):
  h = int(input("Horas trabajadas por el empleado #" + str(i+1) + ": "))
  horas.append(h)
contador = 0
for h in horas:
  if h > 40:
    contador = contador + 1
print("Empleados que trabajaron horas extras: ", contador)
```
